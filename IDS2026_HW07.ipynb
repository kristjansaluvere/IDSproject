**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
## TODO

import matplotlib.pyplot as plt

collisions = pd.read_csv("dft-road-casualty-statistics-collision-2025.csv")

collisions['hour'] = pd.to_datetime(collisions['time'], format='%H:%M', errors='coerce').dt.hour

hourly_counts = collisions['hour'].value_counts().sort_index()

plt.figure(figsize=(10, 5))
hourly_counts.plot(kind='bar')

plt.title('Number of Road Collisions by Hour of Day')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Collisions')
plt.xticks(rotation=0)
plt.show()
```

```python

weather_labels = {
    1: 'Fine',
    2: 'Rain',
    3: 'Snow',
    4: 'Fine + wind',
    5: 'Rain + wind',
    6: 'Snow + wind',
    7: 'Fog or mist',
    8: 'Other'
}

severity_labels = {
    1: 'Fatal',
    2: 'Serious',
    3: 'Slight'
}

data = collisions[collisions['weather_conditions'].isin(weather_labels) & collisions['collision_severity'].isin(severity_labels)].copy()

data['Weather'] = data['weather_conditions'].map(weather_labels)
data['Severity'] = data['collision_severity'].map(severity_labels)

severity_percentages = pd.crosstab(data['Weather'], data['Severity'], normalize='index') * 100

severity_percentages = severity_percentages.reindex(index=list(weather_labels.values()), columns=['Slight', 'Serious', 'Fatal'], fill_value=0)

weather_counts = data['Weather'].value_counts()

severity_percentages.index = [f'{weather} (n={weather_counts.get(weather, 0):,})'for weather in severity_percentages.index]

plt.figure(figsize=(10, 6))

sns.heatmap(severity_percentages, annot=True, fmt='.1f', cmap='YlOrRd', linewidths=0.5, cbar_kws={'label': 'Percentage (%)'})

plt.title('Collision Severity Distribution by Weather Conditions')
plt.xlabel('Collision Severity')
plt.ylabel('Weather Conditions')
plt.tight_layout()
plt.show()
```

```python

speed_counts = (collisions['speed_limit'].value_counts().sort_index())

speed_counts = speed_counts[speed_counts.index > 0]

plt.figure(figsize=(10, 5))

sns.lineplot(x=speed_counts.index, y=speed_counts.values, marker='o', linewidth=2.5)

plt.title('Number of Road Collisions by Speed Limit')
plt.xlabel('Speed Limit (mph)')
plt.ylabel('Number of Collisions')
plt.xticks(speed_counts.index)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Answer:** The number of collisions is lowest during the night. There is noticeable increase at commuting hours which may be related to high traffic volume.  
The highest number of collisions occured on roads with a 30 mph speed limit. This suggests that higher speed limits do not necessarily correspond to higher collision count since the real travelling speed of the vehicle in unknown.  
The distribution of collision severity is relatively similar in all conditions, except snow + windy where the number of collisions recorded is really small (240), which means that this difference should be interpreted cautiously.
